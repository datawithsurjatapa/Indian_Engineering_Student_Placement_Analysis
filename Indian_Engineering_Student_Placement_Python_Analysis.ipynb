```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 8))
```

```python
df["Placement_Status"].value_counts().plot(kind="bar",color='salmon')
plt.title('Placement Distribution')
plt.xlabel('Placement Status')
plt.ylabel('Number of Students')
plt.show()
```

```python
df['Branch'].value_counts().plot(kind='bar',color='teal')
plt.title('Number of Students by Branch')
plt.xlabel('Branch')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns
```

```python
df['CGPA'].hist(bins=30)
plt.title('CGPA Distribution')
plt.xlabel('CGPA')
plt.ylabel('Number of Students')
plt.show()
```

```python
df['CGPA_Level'] = pd.cut(df['CGPA'], bins=[0, 7, 8.5, 10], labels=['Low', 'Medium', 'High'])
```

```python
sns.countplot(x='CGPA_Level', data=df, order=['Low', 'Medium', 'High'], color='steelblue')
plt.title('Number of Students by CGPA Level')
plt.xlabel('CGPA Level')
plt.ylabel('Number of Students')
plt.show()
```

```python
df['Package_LPA'].describe()
sns.boxplot(x=df['Package_LPA'])
```

```python
# CGPA vs placement outcome
sns.boxplot(x='Placement_Status', y='CGPA', data=df,palette='Set3')
```

```python
placement_by_tier = pd.crosstab(df['College_Tier'], df['Placement_Status'], normalize='index')
```

```python
# Placement rate by college tier
placement_by_tier.plot(kind='bar', stacked=True, color=['teal', 'skyblue'])  # matches Not Placed, Placed column order
plt.title('Placement Rate by College Tier')
plt.xlabel('College Tier')
plt.ylabel('Proportion of Students')
plt.legend(title='Placement Status')
plt.xticks(rotation=0)
plt.show()
```

```python
# Count Plot (Pie Chart-like)
plt.figure(figsize=(6, 4))
sns.countplot(x='Gender', data=df, palette='Set2')
plt.title('Count Plot for Gender')
plt.show()
```

```python
# Bar Plot - Grouped Bar Plot for Branch vs. Placement_Status
plt.figure(figsize=(10, 6))
sns.barplot(x='Branch', y='CGPA', hue='Placement_Status', data=df, palette='viridis')
plt.title('Average CGPA by Branch and Placement Status')
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns
corr = df[numeric_cols].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# 7. Violin Plot for Attendance Percentage by Placement Status
plt.figure(figsize=(8, 6))
sns.violinplot(x='Placement_Status', y='Attendance_Percentage', data=df, palette='Pastel1')
plt.title('Attendance Percentage by Placement Status')
plt.show()
```

```python
##Scatter plot: DSA problems solved vs Package earned
sns.scatterplot(x='DSA_Problems_Solved', y='Package_LPA', hue='Placement_Status', data=df, alpha=0.5)
plt.title('DSA Problems Solved vs Package Earned')
plt.xlabel('DSA Problems Solved')
plt.ylabel('Package (LPA)')
plt.legend(title='Placement Status')
plt.show()
```
